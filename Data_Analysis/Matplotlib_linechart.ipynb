```python
import pandas as pd
import matplotlib.pyplot as plt
import random

plt.rcParams['font.sans-serif'] = ['SimHei']  #设置中文字体为黑体

x = [i for i in range(1,25)]
y1 = [random.randint(10,15) for i in x]
y2 = [random.randint(10,30) for i in x]

plt.figure(figsize=(10,5))
# plt.plot(x,y1,y2)
plt.plot(x,y1, label='SYD')
plt.plot(x,y2, label='QLD')

plt.title('温度Temperature Change', fontsize=15)
plt.xlabel('Time')
plt.ylabel('Temp')
plt.xticks(x[1::2])
y_ticks = [i for i in range(5,35)]
plt.yticks(y_ticks)
plt.grid(linestyle = '--', alpha =0.2)
plt.legend(loc='upper right')

plt.show()
```
