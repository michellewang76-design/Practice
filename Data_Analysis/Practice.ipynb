```python
import requests

url = 'https://upload.wikimedia.org/wikipedia/commons/4/41/WenonaCrest-2016.jpg'

# 核心魔法：User-Agent 伪装面具
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

print(f"正在伪装成浏览器下载图片...")

# 把面具 (headers) 一并带上发送请求
response = requests.get(url, headers=headers)

# 养成好习惯：检查服务器是不是真的给了我们 200 (成功) 的状态码
if response.status_code == 200:
    with open('wenona_logo.jpg', 'wb') as f:
        f.write(response.content)
    print("✅ 突破拦截！图片下载成功，快去看看 baidu_fixed.jpy 吧！")
else:
    print(f"❌ 还是被发现了，服务器返回状态码: {response.status_code}")
```
