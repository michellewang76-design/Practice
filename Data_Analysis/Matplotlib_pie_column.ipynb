```python
# import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.axes import Axes

plt.rcParams['font.sans-serif'] = ['SimHei']  #设置中文字体为黑体

figure,axes = plt.subplots(nrows=1, ncols=2, figsize=(20,6), dpi=100)

kids = ['Florence', 'Scarlett', 'Bria', 'Maive']
height = [140, 112, 135, 110]
apple = [6, 4, 2, 1]

axes1: Axes = axes[0]
axes1.bar(kids,height, width=0.6, color='g')
axes1.set_title('Height Chart', fontsize=18)
axes1.set_xlabel('name')
axes1.set_ylabel('cm')
axes1.grid(linestyle='--', alpha=0.4)

axes2: Axes = axes[1]
axes2.pie(apple, labels=kids, autopct='%1.1f%%')
axes2.set_title('Share Apple')
axes2.legend(loc='lower center', ncol=2, bbox_to_anchor=(0.5,-0.1))

plt.show()
```
